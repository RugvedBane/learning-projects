```python
import numpy as np
from tensorflow.keras.layers import Dense, Embedding, SimpleRNN, LSTM, Dropout, Input
from tensorflow.keras.datasets import imdb
from tensorflow.keras import Sequential
from keras.preprocessing.sequence import pad_sequences
```

```python
VOCAB_SIZE = 10000
MAX_LEN = 150

(X_train, y_train), (X_test, y_test) = imdb.load_data(num_words=VOCAB_SIZE)
X_train = pad_sequences(X_train, maxlen=MAX_LEN, truncating='pre')
X_test = pad_sequences(X_test, maxlen=MAX_LEN, truncating='pre')

print(X_train.shape, X_test.shape)
```

## SimpleRNN

```python
rnn_model = Sequential([
    Input(shape=(MAX_LEN,)),
    Embedding(input_dim=VOCAB_SIZE, output_dim=32),
    SimpleRNN(32, return_sequences=False),
    Dropout(0.5),
    Dense(1, activation='sigmoid')
])

rnn_model.summary()
```

```python
rnn_model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

```python
rnn_history = rnn_model.fit(
    X_train, y_train,
    epochs=5,
    batch_size=32,
    validation_data=(X_test, y_test)
)
```

## LSTM

```python
lstm_model = Sequential([
    Input(shape=(MAX_LEN,)),
    Embedding(input_dim=VOCAB_SIZE, output_dim=64),
    LSTM(32, return_sequences=False),
    Dropout(0.5),
    Dense(1, activation='sigmoid')
])

lstm_model.summary()
```

```python
lstm_model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

```python
lstm_history = lstm_model.fit(
    X_train, y_train,
    epochs=5,
    batch_size=32,
    validation_data=(X_test, y_test)
)
```

## Comparison

```python
rnn_best_val = max(rnn_history.history['val_accuracy'])
lstm_best_val = max(lstm_history.history['val_accuracy'])

print(f"SimpleRNN  best val_accuracy: {rnn_best_val:.4f}")
print(f"LSTM       best val_accuracy: {lstm_best_val:.4f}")
print(f"Improvement: +{(lstm_best_val - rnn_best_val):.4f}")
```
